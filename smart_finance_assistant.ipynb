````python

# 🤖 AI Collaboration: Data Loading and Cleaning Function
# SMART FINANCE ASSISTANT


import pandas as pd


# SAMPLE TRANSACTION DATA


sample_transactions = {

    'Date': [
        '2024-08-01',
        '2024-08-02',
        '2024-08-03',
        '2024-08-04'
    ],

    'Amount': [
        '$45.50',
        '$12.00',
        '-$5.00',
        '$89.95'
    ],

    'Category': [
        'Groceries',
        'Transport',
        'Refund',
        'Entertainment'
    ],

    'Description': [
        'Woolworths',
        'Bus Card',
        'Refund Item',
        'Movie Tickets'
    ]
}

# Create sample DataFrame
df_sample = pd.DataFrame(sample_transactions)


# LOAD AND CLEAN FUNCTION


def load_and_clean_transaction_data(file_path):

    """
    Load and clean transaction data.

    Args:
        file_path: CSV file path or DataFrame

    Returns:
        pandas.DataFrame: Cleaned transaction data
    """

    try:


        # LOAD DATA


        if isinstance(file_path, pd.DataFrame):

            df = file_path.copy()

        else:

            df = pd.read_csv(file_path)


        # REQUIRED COLUMNS


        required_columns = [

            'Date',
            'Amount',
            'Category',
            'Description'
        ]

        # Check missing columns
        for col in required_columns:

            if col not in df.columns:

                raise ValueError(
                    f"Missing required column: {col}"
                )


        # REMOVE MISSING VALUES


        df = df.dropna()


        # CLEAN AMOUNT COLUMN


        df['Amount'] = (

            df['Amount']

            .astype(str)

            .str.replace('$', '', regex=False)

            .str.replace(',', '', regex=False)
        )

        # Convert to numeric
        df['Amount'] = pd.to_numeric(

            df['Amount'],

            errors='coerce'
        )

        # Remove invalid numeric rows
        df = df.dropna(subset=['Amount'])


        # CLEAN DATE COLUMN


        df['Date'] = pd.to_datetime(

            df['Date'],

            errors='coerce'
        )

        # Remove invalid dates
        df = df.dropna(subset=['Date'])


        # RESET INDEX


        df = df.reset_index(drop=True)

        print("Transaction data cleaned successfully!")

        return df

    except FileNotFoundError:

        print("Error: CSV file not found.")

    except pd.errors.EmptyDataError:

        print("Error: Uploaded CSV file is empty.")

    except Exception as e:

        print(f"Data Cleaning Error: {e}")


# TEST FUNCTION


clean_data = load_and_clean_transaction_data(
    df_sample
)

print("\nCLEANED TRANSACTION DATA")
print("=" * 40)

print(clean_data)
````

````python

# 🤖 AI Collaboration: Spending Analysis Function
# SMART FINANCE ASSISTANT

import pandas as pd


# SAMPLE CLEAN DATA


sample_data = {

    'Date': [
        '2024-08-01',
        '2024-08-02',
        '2024-08-03',
        '2024-08-04',
        '2024-08-05'
    ],

    'Amount': [
        45.50,
        12.00,
        89.95,
        25.00,
        120.00
    ],

    'Category': [
        'Groceries',
        'Transport',
        'Entertainment',
        'Coffee',
        'Groceries'
    ],

    'Description': [
        'Woolworths',
        'Bus Card',
        'Movie Tickets',
        'Campus Cafe',
        'Coles'
    ]
}

# Create DataFrame
clean_data = pd.DataFrame(sample_data)


# SPENDING ANALYSIS FUNCTION


def analyze_spending_patterns(df):

    """
    Analyze spending patterns and generate business insights.

    Args:
        df: Cleaned transaction DataFrame

    Returns:
        dict: Analysis results and insights
    """


    # USE ONLY POSITIVE SPENDING


    spending_data = df[df['Amount'] > 0]


    # TOTAL SPENDING


    total_spending = round(

        spending_data['Amount'].sum(),

        2
    )


    # AVERAGE TRANSACTION


    average_transaction = round(

        spending_data['Amount'].mean(),

        2
    )


    # CATEGORY TOTALS


    category_totals = (

        spending_data

        .groupby('Category')['Amount']

        .sum()

        .round(2)
    )


    # CATEGORY PERCENTAGES


    category_percentages = round(

        (category_totals / total_spending) * 100,

        2
    )


    # TOP SPENDING CATEGORY


    top_category = category_totals.idxmax()


    # BUSINESS INSIGHTS


    insights = []

    insights.append(
        f"Highest spending category: {top_category}"
    )

    insights.append(
        f"Total spending: ${total_spending}"
    )

    insights.append(
        f"Average transaction amount: ${average_transaction}"
    )

    # Coffee insight
    if 'Coffee' in category_totals.index:

        if category_totals['Coffee'] > 20:

            insights.append(
                "Coffee spending is relatively high."
            )

    # Entertainment insight
    if 'Entertainment' in category_totals.index:

        insights.append(
            "Entertainment expenses should be monitored carefully."
        )

    # STORE RESULTS


    analysis_results = {

        'Total Spending': total_spending,

        'Average Transaction': average_transaction,

        'Category Totals': category_totals,

        'Category Percentages': category_percentages,

        'Top Category': top_category,

        'Business Insights': insights
    }

    return analysis_results


# TEST FUNCTION


analysis = analyze_spending_patterns(clean_data)

print("SPENDING ANALYSIS RESULTS")
print("=" * 40)

for key, value in analysis.items():

    print(f"\n{key}:")

    print(value)
````

````python

# 🛠️ SMART FINANCE ASSISTANT
# CUSTOM SAVINGS CALCULATOR TOOL


# Import agent module
from hands_on_ai import agent


# SAVINGS CALCULATOR FUNCTION


def savings_calculator(
    current_savings,
    monthly_contribution,
    target_amount
):

    """
    Calculate time needed to reach a savings goal.
    """

    try:

        # Convert inputs to numbers
        current_savings = float(current_savings)
        monthly_contribution = float(monthly_contribution)
        target_amount = float(target_amount)


        # VALIDATION


        if current_savings < 0:
            return "Current savings cannot be negative."

        if monthly_contribution <= 0:
            return "Monthly contribution must be greater than 0."

        if target_amount <= 0:
            return "Target amount must be greater than 0."

        if current_savings >= target_amount:
            return "You have already reached your savings goal!"


        # CALCULATIONS


        remaining_amount = (
            target_amount - current_savings
        )

        months_needed = (
            remaining_amount / monthly_contribution
        )

        years_needed = months_needed / 12


        # FORMATTED RESULT


        result = f"""
SAVINGS GOAL REPORT


Current Savings:
${current_savings:.2f}

Monthly Contribution:
${monthly_contribution:.2f}

Target Amount:
${target_amount:.2f}

Remaining Amount:
${remaining_amount:.2f}

Estimated Time to Reach Goal:
{months_needed:.1f} months

Approximate Years:
{years_needed:.1f} years

Keep saving consistently to reach your goal!
"""

        return result

    except Exception as e:

        return f"Error: {e}"


# REGISTER TOOL WITH AGENT SYSTEM


agent.register_tool(
    "savings_calculator",
    "Calculate time required to reach a savings goal",
    savings_calculator
)


# TEST TOOL


test_result = savings_calculator(
    500,
    100,
    2000
)

print(test_result)
````

````python

# 🌐 GRADIO UI INTEGRATION


import gradio as gr
import pandas as pd


# PROCESS CSV FILE


def process_file(file):

    try:

        # Read CSV file
        df = pd.read_csv(file.name)

        # Clean data
        clean_df = load_and_clean_transaction_data(df)

        # Analyze spending
        analysis = analyze_spending_patterns(clean_df)

        # Generate recommendations
        recommendations = generate_financial_recommendations(
            analysis
        )

        # Format result
        result = f"""
TOTAL SPENDING
========================
${analysis['Total Spending']}

AVERAGE TRANSACTION
========================
${analysis['Average Transaction']}

CATEGORY TOTALS
========================
{analysis['Category Totals']}

CATEGORY PERCENTAGES
========================
{analysis['Category Percentages']}

RECOMMENDATIONS
========================
"""

        for rec in recommendations:
            result += f"\n- {rec}"

        return result

    except Exception as e:

        return f"Error: {e}"


# CHATBOT FUNCTION


def finance_chat(message):

    response = create_finance_chat_personality(
        message
    )

    return response


# CREATE GRADIO INTERFACE


def create_finance_assistant_ui():

    with gr.Blocks() as demo:

        gr.Markdown("# 💰 Smart Finance Assistant")


        # CSV ANALYSIS TAB


        with gr.Tab("CSV Analysis"):

            file_input = gr.File(
                label="Upload CSV File"
            )

            output = gr.Textbox(
                label="Financial Analysis",
                lines=20
            )

            analyze_btn = gr.Button(
                "Analyze Spending"
            )

            analyze_btn.click(
                fn=process_file,
                inputs=file_input,
                outputs=output
            )


        # CHATBOT TAB


        with gr.Tab("Finance Chatbot"):

            chatbot = gr.Chatbot()

            msg = gr.Textbox(
                label="Ask a financial question"
            )

            def respond(message, history):

                reply = finance_chat(message)

                history.append(
                    (message, reply)
                )

                return "", history

            msg.submit(
                respond,
                [msg, chatbot],
                [msg, chatbot]
            )

    return demo


# LAUNCH APP


demo = create_finance_assistant_ui()

demo.launch()
````
