# Smart Finance Assistant

## 📖 Project Overview

This project is a Smart Finance Assistant created for the ISYS2001 Final Programming Project.

The application helps users:
- Analyse spending habits
- Understand budgeting patterns
- Generate financial recommendations
- Calculate savings goals
- Interact with an AI finance chatbot

---

## 💻 Technologies Used

- Python
- Pandas
- Gradio
- hands-on-ai
- Google Colab

---

## ✨ Features

- CSV transaction analysis
- Spending summaries by category
- Financial recommendations
- AI-powered finance chatbot
- Savings goal calculator
- Gradio user interface
- Testing and debugging section

---

## 📂 Repository Structure

/data → transaction datasets  
/tests → testing functions  
/ai-conversations → screenshots of AI collaboration  
/docs → pseudocode and planning notes  

---

## 🚀 How to Run

1. Open the notebook in Google Colab
2. Install required libraries
3. Run all notebook cells
4. Launch the Gradio interface
